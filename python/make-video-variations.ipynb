```python
from genericpath import exists
import os

input = 'mov'

if not os.path.exists(input):
    os.mkdir(input)

os.system(f'ffmpeg -i {input}.mov -vcodec hap {input}/{input}-hap.mov')

if not os.path.exists(f'{input}/tiff'):
    os.mkdir(f'{input}/tiff')

os.system(f'ffmpeg -i {input}.mov -vf fps=30 {input}/tiff/{input}%06d.png')
```
