```python
import os

def make_frame_sequence(src,dest):
    if not os.path.exists(dest):
        os.mkdir(dest)

    tiff_file_path = os.path.join(dest,'frame%06d.png')
    os.system(f'ffmpeg -i {src} -vf fps=30 {tiff_file_path}')

def make_video_variations(input):
    # get basename for folders, etc.
    basename = os.path.basename(input).split('.')[0]

    # prefix needed for folder navigation
    to_data_folder = '../bin/data'

    # make basename folder
    if not os.path.exists(os.path.join(to_data_folder,basename)):
        os.mkdir(os.path.join(to_data_folder,basename))

    # full size images
    make_frame_sequence(os.path.join(to_data_folder,input),os.path.join(to_data_folder,basename,'frames'))

    # small video
    cropped_path = os.path.join(to_data_folder,basename,'cropped.mp4')
    if not os.path.exists(cropped_path):
        os.system(f'ffmpeg -i {os.path.join(to_data_folder,input)} -vf crop=ih:ih {cropped_path}')
    else:
        print(f'path exits: {cropped_path}')

    mini_path = os.path.join(to_data_folder,basename,'mini.mp4')
    os.system(f'ffmpeg -i {cropped_path} -vf scale=32:32 {mini_path}')

    os.system(f'rm {cropped_path}')

    # small images
    make_frame_sequence(mini_path,os.path.join(to_data_folder,basename,'mini-frames'))

    # hap video
    hap_path = os.path.join(to_data_folder,basename,'hap.mov')
    if not os.path.exists(hap_path):
        os.system(f'ffmpeg -i {os.path.join(to_data_folder,input)} -vcodec hap {hap_path}')

# file to process
inputs = ['eye4948.mp4']

for input in inputs:
    make_video_variations(input)
```
